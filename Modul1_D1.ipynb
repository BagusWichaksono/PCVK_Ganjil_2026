### Langkah 2: Menghubungkan Google Colab dengan Google Drive
Jalankan sel kode di bawah ini untuk memberi akses Google Colab ke direktori Google Drive.

```python
# Menghubungkan Google Colab ke Google Drive
from google.colab import drive

drive.mount('/content/drive')
```

### Langkah 3: Menyiapkan Folder Kerja di Google Drive
1. Setelah Drive berhasil terhubung (*Mounted at /content/drive*), buka panel **Files** (ikon folder di sebelah kiri).
2. Masuk ke direktori `drive > MyDrive`.
3. Buat folder baru bernama `PCVK` untuk menyimpan seluruh aset/gambar praktikum.

### Langkah 4: Menghubungkan dan Menyimpan Notebook ke GitHub
1. Pada menu navigasi atas Google Colab, klik **File**.
2. Pilih **Save a copy in GitHub**.
3. Jika belum terhubung, lakukan otorisasi akun GitHub kamu.
4. Pilih repositori target (contoh: `PCVK_Genap_2026`).
5. Pilih *branch* (secara bawaan: `main`).
6. Masukkan pesan commit (contoh: `add Modul1_D1.ipynb`).
7. Klik **OK** untuk menyimpan.
